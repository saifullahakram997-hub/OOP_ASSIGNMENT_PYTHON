**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount:
    def __init__(self , balance ):
        self.__balance  = balance
    def deposit(self , deposit_Ammount):
        self.__balance += deposit_Ammount
        print (f"{deposit_Ammount} Successfully Deposited")
    def withdraw(self , withdraw_Ammount):
        if withdraw_Ammount <= self.__balance:
            self.__balance-= withdraw_Ammount
            print(f"{withdraw_Ammount} Withdrawn Sussessfully")
        else:
            print("Insufficent Balance!")
    def show_balance(self):
        print (self.__balance)
```

```python
bank= BankAccount(100000)
```

```python
bank.deposit(50000)
```

```python
bank.withdraw(10000)
```

```python
bank.show_balance()
```

```python
bank.withdraw(200000)
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
class Quantity:
    def __init__(self , value):
        self.value = value
    def __add__(self , other):
        total_value =  self.value + other.value
        return Quantity(total_value)
```

```python
q1 = Quantity(10)
q2 = Quantity(20)
q3 = q1 + q2
print(q3.value)
```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Write your code here for Q15
class compute:
    def area(self, lenght , width=None):
        if width == None:
            return lenght * lenght
        else:
            return lenght * width
        
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
class Payment(ABC):
    @abstractmethod
    def payment(self):
        pass
class CreditCardPayment(Payment):
    def payment(self , ammount):
        print(f"RS.{int(ammount)} payed successfully!")

class UPIPayment(Payment):
    def payment(self , ammount):
        print(f"UPI recived RS.{int(ammount)}")
```

```python
cred= CreditCardPayment()
```

```python
cred.payment(100)
```

```python
up = UPIPayment()
```

```python
up.payment(400)
```
